```python
def init_parameters():
    W1 = np.random.randn(10, 784)
    b1 = np.random.randn(10, 1)
    W2 = np.random.randn(10, 10)
    b2 = np.random.randn(10, 1)
  
    return W1, b1, W2, b2
```

```python
def relu(Z):
    # if elt in Z is <= 0, then elt = 0; else elt = elt
    return np.maximum(0, Z)
```

```python
def softmax(Z):
    return np.exp(Z) / np.sum(np.exp(Z))
```

```python
def forward_propagation(W1, b1, W2, b2, X):
    Z1 = W1.dot(X) + b1
    A1 = relu(Z1)
    Z2 = W2.dot(A1) + b2
    A2 = softmax(Z2)
    
    return Z1, A1, Z2, A2
```

```python
def onehot(Y):
    # classes 0-9
    onehotY = np.zeros((Y.size, 10))

    for i in range(Y.size):
        label = Y[i]
        onehotY[i][label] = 1

    # want each column as an example
    onehotY = onehotY.T
    return onehotY
```

```python
def relu_derivative(Z):
    # > 0 = slope 1 = true
    return Z > 0
```

```python
def backpropagation(Z1, A1, A2, W2, X, Y):
    onehotY = onehot(Y)

    dZ2 = A2 - onehotY
    dW2 = 1 / Y.size * dZ2.dot(A1.T)
    db2 = 1 / Y.size * np.sum(dZ2, axis=1, keepdims=True)

    dZ1 = W2.T.dot(dZ2) * relu_derivative(Z1)
    dW1 = 1 / Y.size * dZ1.dot(X.T)
    db1 = 1 / Y.size * np.sum(dZ1, axis=1, keepdims=True)

    return dW1, db1, dW2, db2
```

```python
def update_parameters(W1, b1, W2, b2, dW1, db1, dW2, db2, alpha):
    W1 = W1 - alpha * dW1
    b1 = b1 - alpha * db1
    W2 = W2 - alpha * dW2
    b2 = b2 - alpha * db2

    return W1, b1, W2, b2
```
